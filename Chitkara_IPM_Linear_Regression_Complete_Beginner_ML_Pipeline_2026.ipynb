# Understanding the Complete Machine Learning Pipeline

## What Was the Objective of This Lab?

The primary objective of this lab is to build a **Machine Learning model** that can predict future **Units Sold (Sales Demand)** using historical business data.

Every day, leading organizations such as **Amazon, Flipkart, Reliance Retail, Walmart, and D-Mart** make thousands of operational decisions based on expected customer demand. Instead of relying only on experience or intuition, these companies analyse historical business data and use Machine Learning models to estimate future sales.

Accurate demand prediction helps organizations answer important business questions such as:

* How many products should be manufactured or ordered?
* How much inventory should be maintained?
* How many employees should be scheduled?
* How much marketing budget should be allocated?
* What revenue can be expected in the coming weeks or months?

The process of estimating future customer demand using historical business data is known as **Demand Forecasting**.

In this lab, we will learn how Machine Learning can help organizations solve this business problem.

The dataset used in this lab is a **cross-functional business dataset**, meaning it contains information collected from multiple departments within an organization. # What Does the Machine Learning Algorithm Learn?

The Machine Learning algorithm does **not memorize the dataset**. Instead, it studies historical business records and learns the relationships between different business variables and sales.

For example, the algorithm tries to discover patterns such as:

* Does increasing marketing expenditure generally increase sales?
* Does increasing product price reduce customer demand?
* Does maintaining sufficient inventory improve sales?
* Does faster product delivery increase customer purchases?
* Do festival seasons generate higher sales than normal periods?

Rather than following manually written rules, the algorithm automatically learns these relationships from historical data and uses them to predict future sales.

---

# Why Did We Clean the Data?

Real-world business data is rarely perfect. It often contains:

* Missing values
* Duplicate records
* Inconsistent information
* Human data-entry errors

For example:

```
Marketing Spend = ?
```

or

```
Order 101
Order 101
```

If these problems are ignored, the Machine Learning model may learn incorrect patterns and produce inaccurate predictions.

Therefore, the first step in every Machine Learning project is **data cleaning**, ensuring that the dataset is complete, accurate, and ready for analysis.

---

# Why Did We Perform Feature Engineering?

The original dataset contains only raw business variables.

Sometimes, combining existing variables creates more meaningful business indicators.

For example, we created:

### Marketing Efficiency

```
Customer Visits
-------------------------
Marketing Spend
```

This feature tells us how effectively the marketing budget generates customer visits.

Instead of only knowing how much money was spent on marketing, the model also learns **how efficiently that investment attracted customers**, making the prediction more informative.

---

# Why Did We Split the Dataset into Training and Testing Data?

Suppose a teacher gives a student **100 practice questions** before an examination.

If the examination contains the same 100 questions, the student may score 100%.

However, this does not prove that the student truly understands the subject.

Machine Learning follows the same principle.

We divide the dataset into two parts:

* **80% Training Data**
* **20% Testing Data**

The model learns patterns from the training data and is then tested on completely unseen data.

This helps us evaluate whether the model can make accurate predictions for new business situations rather than simply remembering historical records.

---

# Why Did We Encode Categorical Variables?

Computers cannot directly understand text such as:

```
North
South
Online
Retail
```

Machine Learning algorithms work only with numerical values.

Therefore, categorical values are converted into numbers using **One-Hot Encoding**.

Example:

| Sales Channel | Encoded Representation |
| ------------- | ---------------------- |
| Online        | 1 0 0                  |
| Retail        | 0 1 0                  |
| Distributor   | 0 0 1                  |

This allows the algorithm to process categorical business information correctly.

---

# Why Did We Scale Numerical Features?

Different numerical variables often have very different ranges.

For example,

```
Marketing Spend = 85,000
```

while

```
Delivery Delay = 3
```

Without scaling, the algorithm may incorrectly assume that variables with larger numerical values are automatically more important.

Scaling transforms numerical variables to a comparable range so that each feature contributes fairly during model training.

---

# Why Did We Build a Machine Learning Pipeline?

In professional Machine Learning projects, preprocessing and model training are not performed as separate disconnected steps.

Instead, they are combined into a **Pipeline**.

A Machine Learning Pipeline automatically performs the following sequence:

```
Raw Data
      ↓
Data Cleaning
      ↓
Encoding
      ↓
Scaling
      ↓
Linear Regression
      ↓
Prediction
```

After creating the pipeline, the complete workflow can be executed using only two commands:

```python
pipeline.fit()
```

and

```python
pipeline.predict()
```

This is considered the standard workflow used in modern Scikit-learn applications.

# What Did the Linear Regression Algorithm Learn?

Linear Regression studies the relationship between multiple business variables and sales.

For example, it may discover patterns such as:

```
Marketing Spend ↑
        ↓
Sales ↑

Product Price ↑
        ↓
Sales ↓

Delivery Delay ↑
        ↓
Sales ↓

Festival Season ↑
        ↓
Sales ↑
```

Using these relationships, the algorithm builds a mathematical model that estimates future sales for new business situations.

---

# How Were Predictions Made?

After training the model, we created a new business scenario.

Example:

```
Marketing Spend = ₹85,000
Product Price = ₹1,500
Inventory Level = 1,500 units
Employee Count = 24
Festival Season = Yes
```

The trained model analyses these business variables and predicts the expected number of units that may be sold.

This prediction represents a future estimate rather than an actual observed value.

---

# Why Did We Evaluate the Model?

Building a model is not enough—we must also determine how accurate it is.

We used three evaluation metrics:

### Mean Absolute Error (MAE)

Measures the average prediction error.

### Root Mean Squared Error (RMSE)

Gives greater importance to larger prediction errors.

### R² Score

Measures how much variation in sales is explained by the model.

For example,

```
R² = 0.91
```

means that the model explains approximately **91% of the variation in sales demand**, indicating strong predictive performance.

## Step 5: Clean the Dataset

````python
# Remove duplicate rows.
data = data.drop_duplicates()

# Create a fresh row index after removing duplicates.
data = data.reset_index(drop=True)

print("Rows after removing duplicates:", data.shape[0])
````

### Output Meaning

Each business record is now represented only once.


````python
# Automatically identify numerical columns.
numerical_columns = data.select_dtypes(
    include=["int64", "float64"]
).columns

# Fill missing numerical values with the median.
# Median is suitable because it is less affected by extreme values.
for column in numerical_columns:
    data[column] = data[column].fillna(
        data[column].median()
    )

print("Numerical missing values handled.")
````

### Output Meaning

Missing numbers have been replaced with the middle value of their respective columns.


````python
# Automatically identify categorical or text columns.
categorical_columns = data.select_dtypes(
    include=["object"]
).columns

# Fill missing categories with the most frequent category.
for column in categorical_columns:
    data[column] = data[column].fillna(
        data[column].mode()[0]
    )

print("Categorical missing values handled.")
````

### Output Meaning

Missing categories have been replaced with the most common category in that column.


````python
# Confirm that cleaning is complete.
print("Remaining missing values:", data.isnull().sum().sum())
print("Remaining duplicate rows:", data.duplicated().sum())
````

### Expected Output

Both values should be `0`. The dataset is now clean and ready for modelling.


## Step 6: Create Simple Business Features

Feature engineering means creating useful new information from existing columns.


````python
# Marketing efficiency shows customer visits generated
# for every ₹1,000 spent on marketing.
data["Marketing_Efficiency"] = (
    data["Customer_Visits"] /
    data["Marketing_Spend"]
) * 1000

# Workforce capacity combines employee count and overtime.
# Eight overtime hours are treated as one additional workday.
data["Workforce_Capacity"] = (
    data["Employee_Count"] +
    data["Overtime_Hours"] / 8
)

# Supply-chain pressure combines supplier time and delivery delay.
data["Supply_Chain_Pressure"] = (
    data["Supplier_Lead_Time_Days"] +
    data["Delivery_Delay_Days"]
)

# Display the new features.
data[
    [
        "Marketing_Efficiency",
        "Workforce_Capacity",
        "Supply_Chain_Pressure"
    ]
].head()
````

## Step 8: Define Inputs and Target

- **Inputs (X):** business factors used to make the prediction
- **Target (y):** the value that the model must predict


````python
# Month_ID is only an identifier, so it is not useful for prediction.
model_data = data.drop(columns=["Month_ID"])

# X contains all input features.
X = model_data.drop(columns=["Units_Sold"])

# y contains the target variable.
y = model_data["Units_Sold"]

print("Input data shape:", X.shape)
print("Target data shape:", y.shape)
````

### Output Meaning

The input table contains the business variables, while the target contains actual units sold.


## Step 9: Split the Dataset

We use:

- **80% training data** to teach the model
- **20% testing data** to check performance on unseen records


````python
# Import the train-test split function.
from sklearn.model_selection import train_test_split

# Split the data into training and testing parts.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
````

### Why This Is Important

Testing on unseen data gives a more realistic estimate of how the model may perform on future business records.


## Step 10: Identify Numerical and Categorical Features

````python
# Select numerical input columns.
model_numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Select categorical input columns.
model_categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical features:")
print(model_numerical_columns)

print("\nCategorical features:")
print(model_categorical_columns)
````

### Why Separate Them?

- Numerical columns are scaled.
- Categorical columns are encoded into machine-readable numbers.


## Step 11: Build the Complete Preprocessing and Model Pipeline

A **Pipeline** keeps preprocessing and model training together. This makes the workflow consistent, reusable, and less prone to data leakage.


````python
# Import preprocessing tools.
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Import Pipeline.
from sklearn.pipeline import Pipeline

# Import the Linear Regression algorithm.
from sklearn.linear_model import LinearRegression

# Scale numerical columns so they are on a comparable scale.
numerical_preprocessing = StandardScaler()

# Convert categories into 0/1 columns.
# handle_unknown="ignore" prevents errors for a new category.
categorical_preprocessing = OneHotEncoder(
    handle_unknown="ignore"
)

# Apply the correct preprocessing to each column type.
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_preprocessing,
            model_numerical_columns
        ),
        (
            "categorical",
            categorical_preprocessing,
            model_categorical_columns
        )
    ]
)

# Combine preprocessing and Linear Regression.
model_pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("model", LinearRegression())
    ]
)

print("Complete Machine Learning pipeline created.")
````

### What the Pipeline Does

1. Scales numerical features.
2. Encodes categorical features.
3. Sends the prepared data to Linear Regression.
4. Applies the same steps during both training and prediction.


## Step 12: Train the Model

````python
# Train the complete pipeline using the training data.
model_pipeline.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
````

### Output Meaning

The model has learned relationships between business inputs and units sold.


## Step 13: Make Predictions

````python
# Predict units sold for the unseen testing records.
y_pred = model_pipeline.predict(X_test)

# Create a small comparison table.
prediction_comparison = pd.DataFrame({
    "Actual_Units_Sold": y_test.values,
    "Predicted_Units_Sold": np.round(y_pred).astype(int)
})

prediction_comparison.head(10)
````

### Output Meaning

The table compares actual sales with model predictions. Smaller differences indicate better predictions.


## Step 14: Evaluate the Model

We use three common regression metrics:

- **MAE:** average prediction error in units
- **RMSE:** gives greater importance to larger errors
- **R²:** proportion of sales variation explained by the model


````python
# Import regression evaluation metrics.
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Calculate Mean Absolute Error.
mae = mean_absolute_error(y_test, y_pred)

# Calculate Root Mean Squared Error.
rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

# Calculate R-squared.
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R-squared (R²):", round(r2, 3))
````

### How to Read the Results

- Lower MAE is better.
- Lower RMSE is better.
- R² closer to 1 is better.
- Model quality must always be judged in relation to the business cost of prediction errors.
